```python
# 2. Modify the central_config.yaml to use 'cuda' instead of 'cpu'
import yaml

with open('central_config.yaml', 'r') as f:
    config = yaml.safe_load(f)

config['runtime']['device'] = 'cuda'

with open('central_config.yaml', 'w') as f:
    yaml.dump(config, f)

print("Config updated to use CUDA!")
```
